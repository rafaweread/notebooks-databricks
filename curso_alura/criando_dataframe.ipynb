```python
dados = [
    ("Produto 1", "Coca Cola", 1),
    ("Produto 2", "Água", 2),
    ("Produto 3", "Balas", 3),
    ("Produto 4", "Bolacha", 4)
]
    
```

```python
df = spark.createDataFrame(
    dados,
    ['categoria', 'descricao', 'identificador']
)

display(df)
```

```python
data = [('Filme A', 'Drama', 120),
        ('Filme B', 'Ação', 95),
        ('Filme C', 'Drama', 110),
        ('Filme D', 'Comédia', 105)]

df = spark.createDataFrame(
    data,
    ['titulo', 'categoria', 'duracao']
)

display(df)
```

```python
df_filtrado = df.filter(df.duracao > 100)
```

```python
display(df_filtrado)
```

```python
df = df.groupBy("categoria").sum('duracao')
```
