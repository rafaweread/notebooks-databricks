### Ler arquivo .CSV em pasta do volume e transformar em dataframe

```python
caminho = "/Volumes/workspace/default/curso_alura/dados-streaming/"
```

```python
df_logs = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(caminho + "logs_streaming.csv")
)
```

```python
display(df_logs.limit(10))
```

```python
df_logs.createOrReplaceTempView("logs")
```

```python
df_logs.printSchema()
```

```python
display(df_logs.select("watch_time_minutes"))
```

```python
from pyspark.sql.functions import col

df_logs = df_logs.withColumn(
    "watch_time_minutes", 
    col("watch_time_minutes").cast("integer")
)
```
