```python
# Importando os tipos de dados necessários para criar o schema
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType

# Dados dos clientes
dados_clientes = [
    (101, "Márcia Silva"),
    (102, "João Costa"),
    (103, "Ana Pereira"),
    (104, "Carlos Souza") # Este cliente não fez pedidos
]
schema_clientes = StructType([
    StructField("id_cliente", IntegerType(), True),
    StructField("nome", StringType(), True)
])

clientes_df = spark.createDataFrame(data=dados_clientes, schema=schema_clientes)

# Dados dos pedidos
dados_pedidos = [
    (1, 101, "Notebook", 4500.00),
    (2, 102, "Mouse", 150.00),
    (3, 101, "Teclado", 250.00),
    (4, 103, "Monitor", 1200.00),
    (5, 999, "Cadeira Gamer", 950.00) # Pedido de um cliente que não está na nossa lista
]
schema_pedidos = StructType([
    StructField("id_pedido", IntegerType(), True),
    StructField("id_cliente", IntegerType(), True),
    StructField("produto", StringType(), True),
    StructField("valor", DoubleType(), True)
])
pedidos_df = spark.createDataFrame(data=dados_pedidos, schema=schema_pedidos)

# Vamos visualizar os DataFrames criados
print("Tabela de Clientes:")
display(clientes_df)

print("Tabela de Pedidos:")
display(pedidos_df)
```

```python
# Inner Join: une apenas clientes que fizeram pedidos e pedidos que têm clientes correspondentes.
inner_join_df = clientes_df.join(pedidos_df, on="id_cliente", how="inner")

print("Resultado do Inner Join:")
display(inner_join_df)
```

```python
# Left Join: queremos todos os clientes, independentemente de terem feito um pedido ou não.
left_join_df = clientes_df.join(pedidos_df, on="id_cliente", how="left")

print("Resultado do Left Join:")
display(left_join_df)
```

```python
# Right Join: queremos todos os pedidos, independentemente de termos o cliente cadastrado.
right_join_df = clientes_df.join(pedidos_df, on="id_cliente", how="right")

print("Resultado do Right Join:")
display(right_join_df)
```

```python
# Full Outer Join: união completa, queremos ver tudo.
full_outer_join_df = clientes_df.join(pedidos_df, on="id_cliente", how="full_outer")

print("Resultado do Full Outer Join:")
display(full_outer_join_df)
```

```python
# Renomeando a coluna em clientes_df para simular a situação
clientes_df_renomeado = clientes_df.withColumnRenamed("id_cliente", "id")

# A condição de join agora é uma expressão booleana
join_condicional = clientes_df_renomeado["id"] == pedidos_df["id_cliente"]

join_com_nomes_diferentes_df = clientes_df_renomeado.join(pedidos_df, on=join_condicional, how="inner")

print("Resultado do Join com nomes de coluna diferentes:")
display(join_com_nomes_diferentes_df)
```
